

Here, we calculate the average earthquake magnitude for each year in the dataset. 
This helps us identify trends and see whether earthquakes are getting stronger or weaker over time. 
We visualize the result using a line chart for clarity.


```python
avg_mag_year = df.groupby('Year')['magnitude'].mean()
avg_mag_year.plot(kind='line', figsize=(10,6), marker='o', color='green')
plt.title('Average Earthquake Magnitude by Year')
plt.xlabel('Year')
plt.ylabel('Average Magnitude')
plt.grid(True)
plt.show()
```



In this step, we explore the relationship between earthquake magnitude and depth. 
A scatter plot is used to visualize whether stronger earthquakes tend to occur deeper or shallower 
in the Earth's crust.


```python
plt.figure(figsize=(10,6))
plt.scatter(df['magnitude'], df['depth'], c='red', alpha=0.6)
plt.title('Earthquake Depth vs Magnitude')
plt.xlabel('Magnitude')
plt.ylabel('Depth (km)')
plt.grid(True)
plt.show()
```



Here, we visualize the geographical distribution of earthquakes using their latitude and longitude. 
The color of each point represents the magnitude, with hotter colors indicating stronger earthquakes. 
This map helps identify regions with high seismic activity.


```python
plt.figure(figsize=(12,6))
plt.scatter(df['longitude'], df['latitude'], c=df['magnitude'], cmap='hot', alpha=0.7)
plt.colorbar(label='Magnitude')
plt.title('Earthquake Locations')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```
